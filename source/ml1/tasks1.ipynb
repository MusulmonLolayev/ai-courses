# Topshiriqlar

1. O'rgatuvchi to'plam bilan test to'plamning nima farqi bor?
2. Agar yangi obyekt bilan o'rgatuvchi to'plamdagi ikkita obyekt o'rtasidagi masofalar teng bo'lib qolsa nima bo'ladi?
3. Nega rasmlarni piksel ko'rinishidan vektor ko'rinishiga o'tkazdik?
4. Biz MNIST rasmlar to'plamini ko'rib chiqdik hamda algoritm 2% xatolik qildi. Bu (xatolik) nima degani?
5. Oldingi darsdagi kodda biz ikki vektor o'rtasidagi masofa sifatida mos piksellarning ayirmasining modulini olib, ularni yig'ib chiqdik. 
    a. Ushbu formula $p(\mathbf{x}, \mathbf{y})$ Chiziqli Algebra 1 qismida o'tilgan ikki vektor orasidagi masofani ifodalaydimi?
    b. Chiziqli Algebra 1 qismida o'tilgan evklid masofa o'lovidan foydalanib, oldingi masalani qayta yozib chiqing.
6. Uyning narxini uning xonalar soni hamda yuzasiga qarab qancha ekanligini baholash qanday turdagi masala hisoblanadi.
7. Shaxslarning qoldirgan xabariga qarab, ushbu xabarni biror guruhga ajaratish masalasi qanday turdagi masala hisoblanadi.
8. Sinflash hamda regressiya masalalari nima uchun "o‘rgatuvchili o‘rganish" hisoblanadi?
9. Aniqlik hamda xatolik ishlatilishi mumkin bo'lgan masalalarga misollar keltiring.
10. Nega muvozanatlashgan baho juda muhim hisoblanadi?
11. Agar sinov to'plamda model juda ham katta xato yoki juda ham kichik aniqlikda ishlayotgan bo'lsa, ushbu modelni ishlatish mumkinmi?
12. Regressiya masalasi uchun keltirilgan ikkita xatolik funksiyalari bilan ikki vektor orasidagi masofani hisoblovchi funksiyalarning qanday farqlari bor?
