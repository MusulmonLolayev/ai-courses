# O‘rnatish

Python dasturlash tilini ishga tushirish uchun biz bir qancha narsalarni mashinaga (kompyuterga) o‘rnatishimiz talab qilinadi. Shulardan birinchisi va eng muhimi Pythonning Interpretatoridir (Tarjimon). Chunki u biz yozgan dastur kodini qayta ishlab mashina tushunadigan tilga tarjima qilib beradi. Bundan tashqari, biz boshqa qo‘shimcha dasturlarga ham muxtoj bo‘lamiz, sababi, ularsiz bevosita Interpretator bilan ishlash bir qancha noqulayliklarni keltirib chiqaradi.

> **Eslatma**. Biz yuqorida Ingliz tilidagi *Interpretator* so‘zini O‘zbek tiliga *Tarjimon* deb o‘girdik. Lekin ushbu  *Tarjimon* so‘zi to‘liq *Interpretatot* so‘zining mazmunini anglatmaydi.  Chunki uning biz yozgan yuqorida darajadagi kodni mashina kodiga to‘g‘iridan-to‘g‘iri o‘girmasdan, u, birinchi, yozgan kodimizni sintaktik tahlil qilib, agar to‘g‘iri bo‘lsa, uni bayt kodiga o‘tkazadi. Bu bayt kod esa Pythonning Virtual mashinasi yordamida haqiqiy dastur ishlagani kabi ishlatiladi.

## Python Interpretatorni o‘rnatish

1. Python.org dan Python binar fayllarini yuklab olish. [Python 3-ning ishonchli versiyasi - Python 3.10.*. (Windows x86)](https://www.python.org/ftp/python/3.10.1/python-3.10.1.exe);
2. [Python 2-ning ishonchli versiyasi - Python 2.7.* (Windows x86-64)](https://www.python.org/ftp/python/2.7.18/python-2.7.18.amd64.msi);
3. O‘rnatish oynasi (rasmda versiya eski):

![O‘rnatish oynasi](./images/python-install.png)

Har doim **Add Python 3.10.* to Path** ga bayroqcha qo‘yishni unutmang. Hamda **Install now** qismini tanglang.


## IDE o‘rnatish

Biz IDE uchun Visual Studio Codeni ishlatamiz. Faqat ushbu muhitni Visual Studio bilan adashtirmang, chunki biz ishlatmochi bo‘lgan muhit, judda ham sodda muhit bo‘lib, faqat bizga kod yozishimizga va qolgan narsalarni esa qo‘shimcha kengaytmalar orqali o‘rnatishimizga imkon beradi va bir qancha dasturlash tillariga moslashadi. U tekin hisoblanib, uni [Visual Studio Code](https://www.code.visualstudio.com) saytidan yuklab olish mumkin. Uni o‘rnatish boshqa dasturlar kabi juda oson va uning asosiy oynasi quyidagicha:

![VS Code oynasi](./images/vs-code.png)

Ushbu muhit bilan ishlash uchun birinchi biror papka ochib olinadi hamda shu papkada ishlanadi. Hamda Python kengaytmasi o‘rnatish har doim tavsiya etiladi.
