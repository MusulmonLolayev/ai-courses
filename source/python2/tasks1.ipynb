# Topshiriqlar 1

`while` operatori yoki umumiy olganda takrorlash operatorlari dasturlashning eng so‘ngi bosqichi hisoblanadi, ya'ni shular orqali ixtiyoriy masalalarni yechish mumkin bo‘ladi. Shu sababdan ham bu yerda ko‘proq ushbu operator uchun topshiriqlar beramiz.

1. Berilgan butun songa bo‘linadigan hamma sonlarni chiqaring.

2. Foydalanuvchi nol bilan tugaydigan ketma-ketlik kiritadi. Ushbu ketma-ketlik ustida quyidagi masalalarni yeching. Ushbu ketma-ketlikdagi hamma:

    a. sonlarning ko‘paytmasini toping.
    b. juft va toq sonlarning sonini toping.
    c. musbat va manifiy sonlarning sonini toping.
    e. 5 va 3 bo‘linadigan sonlarning sonini toping.
    d. ishora almashishlar sonini toping.

3. Bunda butun va qoldiqli bo‘lishdan foydalib, quyidagi masalalarni yeching:
    a. Berilgan butun sonda nechta raqam bor ekanligini aniqlang.
    b. Berilgan butun sonning raqamlar yig‘indisini aniqlang.
    c. Berilgan butun sonni palindrom yoki palindrom emasligini aniqlang.

4. Berilgan n sonigacha bo‘lgan fibanachi sonlarini chop eting.
5. Foydalanuvchi menyusini `while` operatori yordamida hosil qiling. Buning uchun foydalanuvchiga bir nechta menyular ro'yxatini ularning buyruq sonlari bilan birinchi chiqaring, keyin esa foydalanuvchidan shu ro'yxatdagi biror menyuga kirish uchun buyroqning sonini kiritishni so'rang hamda oxirida foydalanuvchiga siz ushbu bo'limdasiz deb bo'lim nomini chop qiling. Agar foydalanuvchi biror noto'g'iri buyruq sonini kiritsa, u holda bu haqda xabar chiqaring hamda foydalanuvchidan qayta buyruq kiritishini so'rang. Ushbu jarayoni foydalanuvchi chiqish buyrug'ini tanlamaguncha davom ettiring.
