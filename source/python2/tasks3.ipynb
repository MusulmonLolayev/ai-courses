# Topshiriqlar 3

Funksiyalarning asosiy vazifasi bu kodni ma'lum qismlarga ajratishdan iboratligini ko'rib o'tdik. Shu sababdan ham, ushbu qismdagi topshiriqlar sifatida bundan oldingi topshiriqlarni funksiya ko'rinishida qayta yozib chiqish deb berdik. Demak, siz aziz o'quvchilar oldingi qismdagi barcha topshiriqlarni funksiya ko'rinishida tasvirlab, ularni ishlatib ko'rasizlar.
