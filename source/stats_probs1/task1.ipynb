# Topshiriqlar 1

1. Universitetda jami 20 ming ta talaba o'qiydi. Ijtimoiy fanlar fakulteti professorlari har bir yo'nalishdan 100 ta talabani tasodifiy tanlab, ularning bergan javoblari asosida so'rovnomani to'ldirishdi. Ushbu tajribadanigi populyatsiya  hamda namunalar miqdorini aniqlang.
2. Yuqoridagi so'rovnomada nimani o'zgaruvchi deb olish mumkin? Namunalar keltiring.
3. Toifali hamda miqdoriy o'zgaruvchilar uchun turli xil misollar keltiring.
4. Yuqoridagi so'rovnoma asosida professorlar talabalarning o'rtacha bo'yining uzunligi 175.7 sm ekanligini hisoblab topishdi. Agar kimdir hamma talabaning bo'ylarining uzunligini topib, shu asosida o'rtacha bo'yini 176.3 sm ekanligini aniqlasa, u holda qaysi son statistika va parametr bo'ladi?
5. Ushbu to'plamlarni beshta son orqali xulosa qiling:
    a. ${4, 8, 6, 7, 12, 3, 74, 66, 25, 85, 34}$;
    b. ${6,  9, 7, 6, 9, 45, 14, 32, 65, 56, 87, 11}$;
    c. ${5, 6, 3, 5, 6, 2, 8, 12, 32, 4, 54, 12, 43, 67, 87}$;
    d. ${9, 0, 12, 4, 5, 2, 12, 45, 22, 5, 4, 6, 8, 43}$.
6. Yuqoridagi to'plamlarning o'rta qiymatini, standart og'ishini hamda farqini hisoblang.
7. Python dasturlash tilida `random` kutubxonasidan foydalanib tasodifiy sonlar hosil qiling, hamda ularni beshta son bilan xulosa qilish dasturini tuzing.
